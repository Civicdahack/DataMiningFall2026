```python
%%html
<h3>Product Catalog</h3>
<div style="margin-bottom: 10px;">
  <button id="filter-btn" onclick="toggleFilter()">Show Under $130</button>
</div>
<div id="products"></div>
<hr>
<div id="cart">Empty</div>
<div>Total: $<span id="total">0.00</span></div>

<script>
  const items = [
    { id: 1, name: 'Wireless Headphones', price: 99.99 },
    { id: 2, name: 'Mechanical Keyboard', price: 129.99 },
    { id: 3, name: 'Smart Watch', price: 199.99 }
  ];

  let cart = {};
  let filterUnder130 = false;

  function toggleFilter() {
    filterUnder130 = !filterUnder130;
    const btn = document.getElementById('filter-btn');
    btn.textContent = filterUnder130 ? 'Show All' : 'Show Under $130';
    render();
  }

  function render() {
    const displayedItems = filterUnder130
      ? items.filter(p => p.price < 130)
      : items;

    document.getElementById('products').innerHTML = displayedItems.map(p => `
      <div>
        ${p.name} - $${p.price.toFixed(2)}
        <button onclick="add(${p.id})">Add</button>
      </div>
    `).join('');

    const cartKeys = Object.keys(cart);
    const cartDiv = document.getElementById('cart');
    const totalSpan = document.getElementById('total');

    if (cartKeys.length === 0) {
      cartDiv.innerHTML = 'Empty';
      totalSpan.textContent = '0.00';
      return;
    }

    let total = 0;
    cartDiv.innerHTML = cartKeys.map(id => {
      const p = items.find(x => x.id == id);
      const qty = cart[id];
      total += p.price * qty;
      return `<div>${p.name} (${qty}) <button onclick="remove(${id})">-</button></div>`;
    }).join('');

    totalSpan.textContent = total.toFixed(2);
  }

  function add(id) {
    cart[id] = (cart[id] || 0) + 1;
    render();
  }

  function remove(id) {
    if (cart[id] > 1) cart[id]--;
    else delete cart[id];
    render();
  }

  render();
</script>
```
